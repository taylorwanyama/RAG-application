```python
df =df.dropna()
print(df.isna().sum().sum())
print(df.shape)
```

```python
df = df.rename(columns={
    #'pe_city':'procuring_entity_city',
    'pe_email': 'procuring_entity_email',
    'pe_telephone': 'procuring_entity_telephone',    
})
```

```python
df.to_csv('active_tenders_cleaned.csv', index=False)
```

```python
documents = []

for _, row in df.iterrows():

    text = f"""
Tender Reference: {row['tender_ref']}

Description: {row['description']}

Procuring Entity: {row['procuring_entity']}

Procurement Method: {row['procurement_method']}

Procurement Category: {row['procurement_category']}

Venue: {row['venue']}

Financial Year: {row['financial_year']}
""".strip()

    metadata = {
        "id": int(row["id"]),
        "tender_ref": row["tender_ref"],
        "procuring_entity": row["procuring_entity"],
        "procurement_method": row["procurement_method"],
        "procurement_category": row["procurement_category"],
        "close_at": row["close_at"],
        "published_at": row["published_at"],
        "addendum_added": int(row["addendum_added"]),
        "addenda_count": int(row["addenda_count"]),
        "venue": row["venue"],
        "tender_fee": row["tender_fee"],
        "validity_in_days": row["validity_in_days"],
        "county_id": row["county_id"],
        "financial_year": row["financial_year"],
        "ocid": row["ocid"],
        "terminated": int(row["terminated"]),
        "tender_document_urls": row["tender_document_urls"],
        "time_to_close": row["time_to_close"],
    }

    documents.append({
        "text": text,
        "metadata": metadata,
    })
```

```python
print(len(documents))
print(documents[0])
```

```python
json.dump(documents, open("tender_documents.json", "w", encoding="utf-8"), indent=2, ensure_ascii=False)
```

```python
print(sum(
    value is None
    for document in documents
    for value in document["metadata"].values()
))
```
